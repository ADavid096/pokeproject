```python
import requests # Para realizar la descarga de info de PokeAPI
import pandas as pd # Para transformar la data
```

## Tabla Pokemon

```python
#Listado de los pokemon segun pokedex global
url_listado_pokemon = "https://pokeapi.co/api/v2/pokemon?limit=10000"
response = requests.get(url_listado_pokemon)
listado_pokemon = response.json()
#Visualizar listado
print(listado_pokemon.keys())
print(listado_pokemon.values())
```

```python
#Actualmente la pokedex va desde el 1 al 1025 (1025)
#Las variantes empiezan a partir de 10000 a 10325 (325)
# 1025 + 325 = 1350
print(listado_pokemon["count"])
print(listado_pokemon["results"])
print(listado_pokemon["results"][0].keys())
print(listado_pokemon["results"][0]["url"].split("/")[-2],":",listado_pokemon["results"][0]["name"])
#Se crea el diccionario que guardara los valores basicos de los pokemon
poke_global = {"poke_id":[],"poke_name":[]}
#Se llena el diccionario con los valores del response
for i in listado_pokemon["results"]:
  poke_global["poke_id"].append(i["url"].split("/")[-2])
  poke_global["poke_name"].append(i["name"])

```

```python
#Se guarda los pokemon en dataframe
import pandas as pd
tabla_pokemon = pd.DataFrame(poke_global)
tabla_pokemon
```

## Tabla Tipos

```python
#Listado de los tipos
url_tipos = "https://pokeapi.co/api/v2/type/?limit=1000"
response = requests.get(url_tipos)
listado_tipos = response.json()
#Visualizar listado
#print(listado_tipos.keys())
#print(listado_tipos.values())

#Se genera un listado de urls para descargar info de los tipos
urls_tipos = []
#Se llena el diccionario con los valores del response
for i in listado_tipos["results"]:
  #tipos["id_tipo"].append(i["url"].split("/")[-2])
  #tipos["nombre_tipo"].append(i["name"])
  urls_tipos.append(i["url"])
urls_tipos
```

```python
#Explorando la info que sale de cada link
response = requests.get(urls_tipos[17])
data = response.json()
#data["damage_relations"]# Efectividad contra otros tipos (USAR EN TABLA DE EFECTIVIDAD TIPO TIPO POR GEN)
# 'double_damage_from', 'double_damage_to', 'half_damage_from', 'half_damage_to', 'no_damage_from', 'no_damage_to'
#data["game_indices"]#(NO USAR) Id en diferentes juegos
#data["generation"]# (USAR) Generacion en la que salio
#data["id"]# (USAR) Id
#data["move_damage_class"]# (USAR) clase de daño de movimiento(de 1era a 3era, los tipos que no son gen 1 a 3 no tienen este dato)
#data["name"]# (USAR) Nombre en ingles
#data["names"]# Nombre en distintos idiomas (NO USAR)
#data["past_damage_relations"]# Efectividad contra tipos de generaciones pasadas (MODIFICACIONES A PARTIR DE X GEN)(USAR EN TABLA DE EFECTIVIDAD TIPO TIPO POR GEN)
#data["pokemon"]# (USAR PARA FORMAPOKEMONGEN) Pokemon con este tipo y el slot en el cual lo tienen presente
#data["sprites"]# (USAR) Imagen por generacion
```

```python
# Desde 1era a 3era Gen el tipo indica que clase de daño se realiza:
# Tipos Especiales: Fuego, Agua, Planta, Eléctrico, Psíquico, Hielo, Dragón y Siniestro.
# Tipos Físicos: Normal, Lucha, Volador, Veneno, Tierra, Roca, Bicho, Fantasma y Acero.
# Tipos que no se usan o para mecanicas: Stellar(Para Teracristalizacion), Shadow(de juegos con otra mecanica), Unknown(no se usa)
```

```python
# Obtener un listado de las claves para los diccionarios para el dataframe
claves = list(data.keys())
claves.pop(claves.index("names"))
claves.pop(claves.index("game_indices"))
claves
```

```python
# Obtener listado de diccionarios con toda la informacion de los tipos.
lista_tipe=[]
for y in range(len(urls_tipos)):
  tipe={}
  response1 = requests.get(urls_tipos[y])
  for clave in claves:
      datax = response1.json()
      tipe[clave] = datax[clave]
  lista_tipe.append(tipe)
  print(y,": ",lista_tipe[y]["name"]," ",lista_tipe[y]["pokemon"])
```

```python
# Tabla final de tipos y Conversion a Dataframe
registrostipos = pd.DataFrame(lista_tipe)
tabla_tipo = registrostipos[["id","name"]].copy()
tabla_tipo["move_damage_class"] = registrostipos["move_damage_class"].str["name"]
tabla_tipo["generation_salida"] = registrostipos["generation"].str["name"]
tabla_tipo
```

```python
# Descargar el dataframe de tipos a CSV
tabla_tipo.to_csv("tablatipos.csv",index=False)
```

## Tabla Generacion

```python
#Listado de los generacion
url_gen = "https://pokeapi.co/api/v2/generation"
response = requests.get(url_gen)
listado_gen = response.json()
x = [] # Array de urls

#Se crea el dataframe para guardar la informacion final
tabla_gen = pd.DataFrame(columns=["num_gen", "gen", "main_region"])

#Se llena el diccionario con los valores del response
for i in listado_gen["results"]:
  x.append(i["url"])
#Se llena el dataframe con las generaciones
for j in range(1,len(x)+1):
  response1 = requests.get(x[j-1])
  data1 = response1.json()
  tabla_gen.loc[len(tabla_gen)] = [j,data1["name"],data1["main_region"]["name"]]

tabla_gen
```

```python
mapa = tabla_gen.set_index("gen")["num_gen"]#Se guarda este diccionario para indicar el "num_gen" en base a la "gen"
```

## Tabla Generacion-Tipo

```python
tipos_general=[]
tipo_json={}
for i in range(1,listado_gen["count"]+1):
  nuevos=[]#Array de diccionarios con los repetidos de la generacion anterior, con el num_gen cambiado
  response1 = requests.get(x[i-1])
  data1 = response1.json()
  print("-----------")
  print(i)#num_gen
  #Se realiza para copiar y agregar los valores repetidos de generaciones antiguas que no son nuevos(La primera vez no aplica)

  if i>1:
    anteriores = [y for y in tipos_general if y["num_gen"] == i - 1]
    print(anteriores)
    for item in anteriores:
        nuevo = item.copy()
        nuevo["num_gen"] = i
        nuevos.append(nuevo)
    tipos_general.extend(nuevos)#Se agrega los datos repetidos de la generacion i
  #Se agrega los tipos que se agregaron en la generacion i
  if(len(data1["types"])>0):
    for j in range(len(data1["types"])):
      tipo_json["num_gen"]=i
      tipo_json["id"]=data1["types"][j]["url"].split("/")[-2]
      tipo_json["name"]=data1["types"][j]["name"]
      tipos_general.append(tipo_json)
      tipo_json={}
#Visualizar el array final
print(tipos_general)

```

```python
tabla_tipos_gen = pd.DataFrame(tipos_general)
tabla_tipos_gen
```

```python
#Obtener los sprites por tipo
gencolector = []
for url in urls_tipos:
  response = requests.get(url)
  data =response.json()
  eldatas = pd.DataFrame(data["sprites"])
  sprites_url = list(eldatas.keys())#Listado de encabezados ("generation-i","generation-ii",etc)

  #Se crea el listado de diccionarios para el tipo en el listado de urls
  for c in range(len(sprites_url)):
    gensprite = {}

    eldotas = eldatas[sprites_url[c]].dropna()#Se borran todos los valores NaN
    gensprite["id"] = data["id"]
    gensprite["name"] = data["name"]
    gensprite["gen"] = sprites_url[c]#Generacion
    gensprite["spriteurl"] = eldotas.iloc[0]["name_icon"]#Imagen de la generacion
    if c==0:
      url_valor=gensprite["spriteurl"]
    gencolector.append(gensprite)

  #Agregar la primera y segunda generacion
  gencolector.append({"id":data["id"],"name":data["name"],"gen":"generation-i","spriteurl":url_valor})
  gencolector.append({"id":data["id"],"name":data["name"],"gen":"generation-ii","spriteurl":url_valor})

#Visualizar resultado
#Este dataframe bota el id tipo, el nombre del tipo y la generacion en texto y el sprite
dfgencolector =pd.DataFrame(gencolector)
#Se agrega el num_gen al final, luego de realizar el dataframe final con todos los tipos
dfgencolector["num_gen"] = dfgencolector["gen"].map(mapa)
dfgencolector
```

```python
#Se agrega una columna al dfgencolector y se pasa a index
dfgencolector2 = dfgencolector.copy()
dfgencolector2["num_gen-id"] = "gen-" + dfgencolector2["num_gen"].astype(str) + "-type-" + dfgencolector2["id"].astype(str)
dfgencolector2 = dfgencolector2.set_index("num_gen-id")
dfgencolector2.to_csv("pruebas.csv")
```

```python
mapa_gen = dfgencolector2["gen"]#Serie diccionario para "gen"
mapa_spriteurl = dfgencolector2["spriteurl"]#Serie diccionario para "spriteurl"

mapa_spriteurl.to_csv("mapa_spriteurl.csv")
mapa_gen.to_csv("mapa_gen.csv")
```

```python
#Se crea el nuevo dataframe con el nombre de la gen y el sprite (Tabla final Tipo-Gen)
tabla_tipos_gen2 = tabla_tipos_gen.copy()
tabla_tipos_gen2["combina"] = tabla_tipos_gen2["num_gen"].astype(str) + "-" + tabla_tipos_gen2["id"].astype(str)
tabla_tipos_gen2
#Se agrega en base a los diccionarios de "mapa_gen" y "mapa_spriteurl"
tabla_tipos_gen2["gen"] = tabla_tipos_gen2["combina"].map(mapa_gen)
tabla_tipos_gen2["spriteurl"] = tabla_tipos_gen2["combina"].map(mapa_spriteurl)
#Ver resultado final
tabla_tipos_gen2.head(40)
```
